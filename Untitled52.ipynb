```python
import os
import json
import pickle
import zipfile
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import (
    silhouette_score,
    adjusted_rand_score,
    davies_bouldin_score
)

from scipy.stats import f_oneway

try:
    from minisom import MiniSom
except ImportError:
    raise ImportError(
        "MiniSom is not installed.\n"
        "Run: pip install minisom"
    )

warnings.filterwarnings("ignore")


# =====================================================================
# 2. PROJECT PATHS
# =====================================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads\Self-Organizing Maps for Clustering"
)

DATA_PATH = PROJECT_DIR / "CC GENERAL.csv"

RESULTS_DIR = PROJECT_DIR / "results"
GRAPHS_DIR = RESULTS_DIR / "graphs"
MODELS_DIR = RESULTS_DIR / "models"
DATA_DIR = RESULTS_DIR / "data"
REPORT_DIR = RESULTS_DIR / "reports"

for folder in [
    RESULTS_DIR,
    GRAPHS_DIR,
    MODELS_DIR,
    DATA_DIR,
    REPORT_DIR
]:
    folder.mkdir(parents=True, exist_ok=True)

print("=" * 70)
print("SELF-ORGANIZING MAP CUSTOMER SEGMENTATION")
print("=" * 70)

print("\nProject directory:")
print(PROJECT_DIR)

print("\nDataset:")
print(DATA_PATH)


# =====================================================================
# 3. CONFIGURATION
# =====================================================================

RANDOM_STATE = 42

SOM_X = 10
SOM_Y = 10

SOM_SIGMA = 3.0
SOM_LEARNING_RATE = 0.5

SOM_ITERATIONS = 10000

NUMBER_OF_CLUSTERS = 4

np.random.seed(RANDOM_STATE)


# =====================================================================
# 4. HELPER FUNCTION FOR SAVING GRAPHS
# =====================================================================

def save_plot(filename):
    path = GRAPHS_DIR / filename

    plt.tight_layout()

    plt.savefig(
        path,
        dpi=180,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()

    print("Saved:", path)


# =====================================================================
# 5. LOAD DATASET
# =====================================================================

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATA_PATH}"
    )

df = pd.read_csv(DATA_PATH)

print("\nDataset loaded successfully.")

print("\nDataset shape:")
print(df.shape)

print("\nFirst five rows:")
print(df.head())

print("\nColumns:")
print(df.columns.tolist())


# =====================================================================
# 6. BASIC DATA INFORMATION
# =====================================================================

print("\n" + "=" * 70)
print("DATA INFORMATION")
print("=" * 70)

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())


# =====================================================================
# 7. SAVE ORIGINAL DATA SUMMARY
# =====================================================================

summary = df.describe(include="all").transpose()

summary.to_csv(
    DATA_DIR / "original_data_summary.csv"
)

print(
    "\nSaved:",
    DATA_DIR / "original_data_summary.csv"
)


# =====================================================================
# 8. REMOVE DUPLICATES
# =====================================================================

df = df.drop_duplicates().reset_index(drop=True)


# =====================================================================
# 9. CUSTOMER ID HANDLING
# =====================================================================

if "CUST_ID" in df.columns:

    customer_ids = df["CUST_ID"].copy()

    features_df = df.drop(
        columns=["CUST_ID"]
    ).copy()

else:

    customer_ids = pd.Series(
        np.arange(len(df)),
        name="CUST_ID"
    )

    features_df = df.copy()


# =====================================================================
# 10. HANDLE MISSING VALUES
# =====================================================================

numeric_columns = features_df.select_dtypes(
    include=np.number
).columns.tolist()

for column in numeric_columns:

    median_value = features_df[column].median()

    features_df[column] = features_df[column].fillna(
        median_value
    )

print("\nMissing values after median imputation:")
print(features_df.isnull().sum().sum())


# =====================================================================
# 11. FEATURE ENGINEERING
# =====================================================================

epsilon = 1e-8

if (
    "PURCHASES" in features_df.columns
    and
    "CREDIT_LIMIT" in features_df.columns
):

    features_df[
        "PURCHASES_CREDIT_RATIO"
    ] = (
        features_df["PURCHASES"]
        /
        (
            features_df["CREDIT_LIMIT"]
            + epsilon
        )
    )


if (
    "PAYMENTS" in features_df.columns
    and
    "CREDIT_LIMIT" in features_df.columns
):

    features_df[
        "PAYMENT_CREDIT_RATIO"
    ] = (
        features_df["PAYMENTS"]
        /
        (
            features_df["CREDIT_LIMIT"]
            + epsilon
        )
    )


if (
    "BALANCE" in features_df.columns
    and
    "CREDIT_LIMIT" in features_df.columns
):

    features_df[
        "BALANCE_CREDIT_RATIO"
    ] = (
        features_df["BALANCE"]
        /
        (
            features_df["CREDIT_LIMIT"]
            + epsilon
        )
    )


if (
    "CASH_ADVANCE" in features_df.columns
    and
    "CREDIT_LIMIT" in features_df.columns
):

    features_df[
        "CASH_ADVANCE_CREDIT_RATIO"
    ] = (
        features_df["CASH_ADVANCE"]
        /
        (
            features_df["CREDIT_LIMIT"]
            + epsilon
        )
    )


if (
    "PURCHASES" in features_df.columns
    and
    "PAYMENTS" in features_df.columns
):

    features_df[
        "PURCHASE_PAYMENT_RATIO"
    ] = (
        features_df["PURCHASES"]
        /
        (
            features_df["PAYMENTS"]
            + epsilon
        )
    )


print("\nFeatures after engineering:")
print(features_df.columns.tolist())

print("\nNumber of features:")
print(features_df.shape[1])


# =====================================================================
# 12. REMOVE INF / NAN CREATED BY RATIOS
# =====================================================================

features_df = features_df.replace(
    [np.inf, -np.inf],
    np.nan
)

for column in features_df.columns:

    if features_df[column].isnull().any():

        features_df[column] = (
            features_df[column]
            .fillna(features_df[column].median())
        )


# =====================================================================
# 13. FEATURE DISTRIBUTION HISTOGRAMS
# =====================================================================

feature_columns = features_df.columns.tolist()

n_features = len(feature_columns)

ncols = 4
nrows = int(np.ceil(n_features / ncols))

fig, axes = plt.subplots(
    nrows,
    ncols,
    figsize=(18, 4 * nrows)
)

axes = np.array(axes).flatten()

for i, column in enumerate(feature_columns):

    axes[i].hist(
        features_df[column],
        bins=30,
        edgecolor="black",
        alpha=0.75
    )

    axes[i].set_title(column)

    axes[i].set_xlabel("Value")
    axes[i].set_ylabel("Frequency")

for j in range(i + 1, len(axes)):
    axes[j].axis("off")

plt.suptitle(
    "Feature Distributions",
    fontsize=16,
    y=1.01
)

save_plot(
    "01_feature_distributions.png"
)


# =====================================================================
# 14. FEATURE BOX PLOTS
# =====================================================================

fig, axes = plt.subplots(
    nrows,
    ncols,
    figsize=(18, 4 * nrows)
)

axes = np.array(axes).flatten()

for i, column in enumerate(feature_columns):

    axes[i].boxplot(
        features_df[column].dropna()
    )

    axes[i].set_title(column)

for j in range(i + 1, len(axes)):
    axes[j].axis("off")

plt.suptitle(
    "Feature Box Plots",
    fontsize=16,
    y=1.01
)

save_plot(
    "02_feature_boxplots.png"
)


# =====================================================================
# 15. CORRELATION HEATMAP
# =====================================================================

correlation_matrix = features_df.corr()

fig, ax = plt.subplots(
    figsize=(18, 15)
)

image = ax.imshow(
    correlation_matrix,
    cmap="coolwarm",
    aspect="auto",
    vmin=-1,
    vmax=1
)

ax.set_xticks(
    np.arange(len(feature_columns))
)

ax.set_yticks(
    np.arange(len(feature_columns))
)

ax.set_xticklabels(
    feature_columns,
    rotation=90,
    fontsize=7
)

ax.set_yticklabels(
    feature_columns,
    fontsize=7
)

plt.colorbar(
    image,
    ax=ax,
    label="Correlation"
)

plt.title(
    "Feature Correlation Heatmap"
)

save_plot(
    "03_correlation_heatmap.png"
)


# =====================================================================
# 16. IQR OUTLIER DETECTION
# =====================================================================

outlier_summary = []

for column in feature_columns:

    q1 = features_df[column].quantile(0.25)
    q3 = features_df[column].quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = (
        (features_df[column] < lower_bound)
        |
        (features_df[column] > upper_bound)
    )

    outlier_summary.append({
        "Feature": column,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": int(outliers.sum()),
        "Outlier_Percentage":
            float(outliers.mean() * 100)
    })


outlier_df = pd.DataFrame(
    outlier_summary
)

outlier_df.to_csv(
    DATA_DIR / "outlier_analysis.csv",
    index=False
)

print("\nOutlier analysis:")
print(
    outlier_df[
        [
            "Feature",
            "Outlier_Count",
            "Outlier_Percentage"
        ]
    ]
)


# =====================================================================
# 17. OPTIONAL OUTLIER CAPPING
#
# Credit-card spending data naturally contains extreme customers.
# Instead of deleting customers, values are winsorized using IQR bounds.
# =====================================================================

capped_df = features_df.copy()

for column in feature_columns:

    q1 = capped_df[column].quantile(0.25)
    q3 = capped_df[column].quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    capped_df[column] = capped_df[column].clip(
        lower=lower_bound,
        upper=upper_bound
    )


# =====================================================================
# 18. MIN-MAX NORMALIZATION
# =====================================================================

minmax_scaler = MinMaxScaler()

X_minmax = minmax_scaler.fit_transform(
    capped_df
)


# =====================================================================
# 19. Z-SCORE STANDARDIZATION
# =====================================================================

standard_scaler = StandardScaler()

X_standard = standard_scaler.fit_transform(
    capped_df
)


print("\nScaling completed.")

print(
    "MinMax range:",
    X_minmax.min(),
    "to",
    X_minmax.max()
)

print(
    "Standardized mean:",
    np.mean(X_standard)
)


# =====================================================================
# 20. SAVE SCALERS
# =====================================================================

with open(
    MODELS_DIR / "minmax_scaler.pkl",
    "wb"
) as file:

    pickle.dump(
        minmax_scaler,
        file,
        protocol=pickle.HIGHEST_PROTOCOL
    )


with open(
    MODELS_DIR / "standard_scaler.pkl",
    "wb"
) as file:

    pickle.dump(
        standard_scaler,
        file,
        protocol=pickle.HIGHEST_PROTOCOL
    )


# =====================================================================
# 21. PCA 2D
# =====================================================================

pca_2d = PCA(
    n_components=2,
    random_state=RANDOM_STATE
)

X_pca_2d = pca_2d.fit_transform(
    X_standard
)

plt.figure(
    figsize=(10, 7)
)

plt.scatter(
    X_pca_2d[:, 0],
    X_pca_2d[:, 1],
    s=15,
    alpha=0.6
)

plt.xlabel(
    f"PC1 ({pca_2d.explained_variance_ratio_[0] * 100:.2f}%)"
)

plt.ylabel(
    f"PC2 ({pca_2d.explained_variance_ratio_[1] * 100:.2f}%)"
)

plt.title(
    "PCA 2D Projection Before SOM Clustering"
)

plt.grid(alpha=0.2)

save_plot(
    "04_pca_2d_before_clustering.png"
)


# =====================================================================
# 22. COMPARE MINMAX VS STANDARD SCALING USING PCA
# =====================================================================

pca_minmax = PCA(
    n_components=2,
    random_state=RANDOM_STATE
)

pca_standard = PCA(
    n_components=2,
    random_state=RANDOM_STATE
)

minmax_pca = pca_minmax.fit_transform(
    X_minmax
)

standard_pca = pca_standard.fit_transform(
    X_standard
)

scaling_comparison = pd.DataFrame({
    "Scaling": [
        "MinMax",
        "Standard"
    ],
    "PC1_Variance": [
        pca_minmax.explained_variance_ratio_[0],
        pca_standard.explained_variance_ratio_[0]
    ],
    "PC2_Variance": [
        pca_minmax.explained_variance_ratio_[1],
        pca_standard.explained_variance_ratio_[1]
    ]
})

scaling_comparison[
    "Total_2D_Variance"
] = (
    scaling_comparison["PC1_Variance"]
    +
    scaling_comparison["PC2_Variance"]
)

scaling_comparison.to_csv(
    DATA_DIR / "scaling_comparison.csv",
    index=False
)

print("\nScaling comparison:")
print(scaling_comparison)


# =====================================================================
# 23. SOM TRAINING DATA
#
# MinMax is used because the assignment explicitly requires [0,1]
# normalization and SOM distance calculations are scale sensitive.
# =====================================================================

X = X_minmax.astype(
    np.float32
)

print(
    "\nSOM input shape:",
    X.shape
)


# =====================================================================
# 24. CREATE SOM
# =====================================================================

som = MiniSom(
    x=SOM_X,
    y=SOM_Y,
    input_len=X.shape[1],
    sigma=SOM_SIGMA,
    learning_rate=SOM_LEARNING_RATE,
    neighborhood_function="gaussian",
    topology="rectangular",
    activation_distance="euclidean",
    random_seed=RANDOM_STATE
)


# =====================================================================
# 25. PCA WEIGHT INITIALIZATION
# =====================================================================

print(
    "\nInitializing SOM weights using PCA..."
)

som.pca_weights_init(X)


# =====================================================================
# 26. TRACK QUANTIZATION ERROR WHILE TRAINING
#
# Training is performed in blocks. After each block, the quantization
# error is calculated.
# =====================================================================

print("\nTraining SOM...")

error_history = []
iteration_history = []

CHECK_INTERVAL = 500

completed_iterations = 0

while completed_iterations < SOM_ITERATIONS:

    current_iterations = min(
        CHECK_INTERVAL,
        SOM_ITERATIONS - completed_iterations
    )

    som.train_random(
        X,
        current_iterations,
        verbose=False
    )

    completed_iterations += current_iterations

    qe = som.quantization_error(X)

    error_history.append(qe)

    iteration_history.append(
        completed_iterations
    )

    print(
        f"Iteration "
        f"{completed_iterations:5d}/{SOM_ITERATIONS} "
        f"| Quantization Error: {qe:.6f}"
    )


print("\nSOM training completed.")


# =====================================================================
# 27. QUANTIZATION ERROR GRAPH
# =====================================================================

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    iteration_history,
    error_history,
    marker="o"
)

plt.xlabel(
    "Training Iterations"
)

plt.ylabel(
    "Quantization Error"
)

plt.title(
    "SOM Quantization Error Convergence"
)

plt.grid(alpha=0.3)

save_plot(
    "05_quantization_error_convergence.png"
)


quantization_df = pd.DataFrame({
    "Iteration": iteration_history,
    "Quantization_Error": error_history
})

quantization_df.to_csv(
    DATA_DIR / "quantization_error_history.csv",
    index=False
)


# =====================================================================
# 28. SAVE SOM WEIGHTS
# =====================================================================

som_weights = som.get_weights().astype(
    np.float32
)

np.savez_compressed(
    MODELS_DIR / "som_weights.npz",
    weights=som_weights
)

print(
    "\nSOM weights saved."
)


# =====================================================================
# 29. SAVE SOM MODEL
# =====================================================================

with open(
    MODELS_DIR / "trained_som.pkl",
    "wb"
) as file:

    pickle.dump(
        som,
        file,
        protocol=pickle.HIGHEST_PROTOCOL
    )


# =====================================================================
# 30. U-MATRIX
# =====================================================================

u_matrix = som.distance_map()

plt.figure(
    figsize=(9, 8)
)

plt.imshow(
    u_matrix.T,
    origin="lower",
    cmap="bone_r"
)

plt.colorbar(
    label="Average Neighbor Distance"
)

plt.xlabel(
    "SOM X Coordinate"
)

plt.ylabel(
    "SOM Y Coordinate"
)

plt.title(
    "SOM U-Matrix"
)

save_plot(
    "06_u_matrix.png"
)


# =====================================================================
# 31. HIT MAP
# =====================================================================

hit_map = np.zeros(
    (SOM_X, SOM_Y),
    dtype=int
)

bmu_coordinates = []

for row in X:

    winner = som.winner(row)

    bmu_coordinates.append(
        winner
    )

    hit_map[winner] += 1


plt.figure(
    figsize=(9, 8)
)

plt.imshow(
    hit_map.T,
    origin="lower",
    cmap="viridis"
)

plt.colorbar(
    label="Number of Customers"
)

plt.xlabel(
    "SOM X Coordinate"
)

plt.ylabel(
    "SOM Y Coordinate"
)

plt.title(
    "SOM Hit Map"
)

for x in range(SOM_X):

    for y in range(SOM_Y):

        if hit_map[x, y] > 0:

            plt.text(
                x,
                y,
                str(hit_map[x, y]),
                ha="center",
                va="center",
                fontsize=7
            )

save_plot(
    "07_hit_map.png"
)


# =====================================================================
# 32. COMPONENT PLANES
# =====================================================================

component_dir = (
    GRAPHS_DIR /
    "component_planes"
)

component_dir.mkdir(
    exist_ok=True
)

for index, feature in enumerate(
    feature_columns
):

    plane = som_weights[
        :,
        :,
        index
    ]

    plt.figure(
        figsize=(8, 7)
    )

    plt.imshow(
        plane.T,
        origin="lower",
        cmap="coolwarm"
    )

    plt.colorbar(
        label="Normalized Feature Value"
    )

    plt.xlabel(
        "SOM X Coordinate"
    )

    plt.ylabel(
        "SOM Y Coordinate"
    )

    plt.title(
        f"Component Plane - {feature}"
    )

    plt.tight_layout()

    safe_name = (
        feature
        .replace("/", "_")
        .replace("\\", "_")
        .replace(" ", "_")
    )

    plt.savefig(
        component_dir /
        f"component_{safe_name}.png",
        dpi=150,
        bbox_inches="tight"
    )

    plt.close()


print(
    "\nComponent planes saved:",
    component_dir
)


# =====================================================================
# 33. CLUSTER SOM NEURONS USING K-MEANS
# =====================================================================

flattened_weights = som_weights.reshape(
    SOM_X * SOM_Y,
    X.shape[1]
)

som_kmeans = KMeans(
    n_clusters=NUMBER_OF_CLUSTERS,
    random_state=RANDOM_STATE,
    n_init=20
)

neuron_cluster_labels = (
    som_kmeans.fit_predict(
        flattened_weights
    )
)

som_cluster_map = (
    neuron_cluster_labels.reshape(
        SOM_X,
        SOM_Y
    )
)


# =====================================================================
# 34. CLUSTER-LABELED SOM MAP
# =====================================================================

plt.figure(
    figsize=(9, 8)
)

plt.imshow(
    som_cluster_map.T,
    origin="lower",
    cmap="tab10"
)

plt.colorbar(
    label="Cluster"
)

plt.xlabel(
    "SOM X Coordinate"
)

plt.ylabel(
    "SOM Y Coordinate"
)

plt.title(
    "SOM Map Colored by Cluster"
)

for x in range(SOM_X):

    for y in range(SOM_Y):

        plt.text(
            x,
            y,
            str(
                som_cluster_map[x, y] + 1
            ),
            ha="center",
            va="center",
            fontsize=7
        )

save_plot(
    "08_cluster_labeled_som_map.png"
)


# =====================================================================
# 35. MAP EACH CUSTOMER TO BMU + CLUSTER
# =====================================================================

customer_clusters = []

bmu_x = []
bmu_y = []

for row in X:

    winner = som.winner(row)

    x_coordinate = winner[0]
    y_coordinate = winner[1]

    cluster = som_cluster_map[
        x_coordinate,
        y_coordinate
    ]

    bmu_x.append(
        x_coordinate
    )

    bmu_y.append(
        y_coordinate
    )

    customer_clusters.append(
        cluster
    )


customer_clusters = np.array(
    customer_clusters
)


# =====================================================================
# 36. CUSTOMER RESULTS DATAFRAME
# =====================================================================

results_df = df.copy()

results_df["BMU_X"] = bmu_x
results_df["BMU_Y"] = bmu_y

results_df["SOM_CLUSTER"] = (
    customer_clusters + 1
)

results_df.to_csv(
    DATA_DIR /
    "customer_cluster_assignments.csv",
    index=False
)

print(
    "\nCustomer cluster assignments saved."
)


# =====================================================================
# 37. SAVE PREPROCESSED DATA
# =====================================================================

preprocessed_df = capped_df.copy()

preprocessed_df.insert(
    0,
    "CUST_ID",
    customer_ids.values
)

preprocessed_df[
    "SOM_CLUSTER"
] = customer_clusters + 1

preprocessed_df.to_csv(
    DATA_DIR /
    "preprocessed_data_with_clusters.csv",
    index=False
)


# =====================================================================
# 38. PCA SCATTER COLORED BY SOM CLUSTER
# =====================================================================

plt.figure(
    figsize=(10, 7)
)

scatter = plt.scatter(
    X_pca_2d[:, 0],
    X_pca_2d[:, 1],
    c=customer_clusters,
    cmap="tab10",
    s=18,
    alpha=0.7
)

plt.colorbar(
    scatter,
    label="SOM Cluster"
)

plt.xlabel(
    f"PC1 ({pca_2d.explained_variance_ratio_[0] * 100:.2f}%)"
)

plt.ylabel(
    f"PC2 ({pca_2d.explained_variance_ratio_[1] * 100:.2f}%)"
)

plt.title(
    "PCA Projection Colored by SOM Cluster"
)

plt.grid(alpha=0.2)

save_plot(
    "09_pca_som_clusters.png"
)


# =====================================================================
# 39. 3D PCA VISUALIZATION
# =====================================================================

pca_3d = PCA(
    n_components=3,
    random_state=RANDOM_STATE
)

X_pca_3d = pca_3d.fit_transform(
    X_standard
)

fig = plt.figure(
    figsize=(11, 8)
)

ax = fig.add_subplot(
    111,
    projection="3d"
)

scatter = ax.scatter(
    X_pca_3d[:, 0],
    X_pca_3d[:, 1],
    X_pca_3d[:, 2],
    c=customer_clusters,
    cmap="tab10",
    s=15,
    alpha=0.65
)

ax.set_xlabel(
    "Principal Component 1"
)

ax.set_ylabel(
    "Principal Component 2"
)

ax.set_zlabel(
    "Principal Component 3"
)

ax.set_title(
    "3D PCA Visualization of SOM Customer Segments"
)

fig.colorbar(
    scatter,
    ax=ax,
    label="SOM Cluster",
    shrink=0.6
)

save_plot(
    "10_pca_3d_clusters.png"
)


# =====================================================================
# 40. CLUSTER SIZE
# =====================================================================

cluster_counts = pd.Series(
    customer_clusters + 1
).value_counts().sort_index()

cluster_percentage = (
    cluster_counts
    /
    len(customer_clusters)
    *
    100
)

cluster_size_df = pd.DataFrame({
    "Cluster": cluster_counts.index,
    "Customer_Count": cluster_counts.values,
    "Percentage": cluster_percentage.values
})

cluster_size_df.to_csv(
    DATA_DIR /
    "cluster_sizes.csv",
    index=False
)

print("\nCluster sizes:")
print(cluster_size_df)


# =====================================================================
# 41. CLUSTER SIZE GRAPH
# =====================================================================

plt.figure(
    figsize=(9, 6)
)

plt.bar(
    cluster_size_df["Cluster"].astype(str),
    cluster_size_df["Customer_Count"]
)

plt.xlabel(
    "SOM Cluster"
)

plt.ylabel(
    "Number of Customers"
)

plt.title(
    "Customer Distribution Across SOM Clusters"
)

for index, row in cluster_size_df.iterrows():

    plt.text(
        index,
        row["Customer_Count"],
        f'{row["Percentage"]:.1f}%',
        ha="center",
        va="bottom"
    )

save_plot(
    "11_cluster_size_distribution.png"
)


# =====================================================================
# 42. CLUSTER MEAN PROFILE
# =====================================================================

profile_data = capped_df.copy()

profile_data[
    "Cluster"
] = customer_clusters + 1

cluster_means = (
    profile_data
    .groupby("Cluster")
    .mean()
)

cluster_std = (
    profile_data
    .groupby("Cluster")
    .std()
)

cluster_means.to_csv(
    DATA_DIR /
    "cluster_feature_means.csv"
)

cluster_std.to_csv(
    DATA_DIR /
    "cluster_feature_std.csv"
)


# =====================================================================
# 43. NORMALIZED CLUSTER PROFILE FOR RADAR CHART
# =====================================================================

radar_scaler = MinMaxScaler()

radar_values = radar_scaler.fit_transform(
    cluster_means
)

radar_df = pd.DataFrame(
    radar_values,
    index=cluster_means.index,
    columns=cluster_means.columns
)

radar_df.to_csv(
    DATA_DIR /
    "cluster_normalized_profiles.csv"
)


# =====================================================================
# 44. RADAR CHART
#
# Use a manageable number of highly variable features.
# =====================================================================

feature_variability = (
    radar_df.std()
    .sort_values(
        ascending=False
    )
)

radar_features = (
    feature_variability
    .head(10)
    .index
    .tolist()
)

radar_subset = radar_df[
    radar_features
]

num_vars = len(
    radar_features
)

angles = np.linspace(
    0,
    2 * np.pi,
    num_vars,
    endpoint=False
).tolist()

angles += angles[:1]

fig, ax = plt.subplots(
    figsize=(10, 10),
    subplot_kw={
        "polar": True
    }
)

for cluster in radar_subset.index:

    values = (
        radar_subset
        .loc[cluster]
        .tolist()
    )

    values += values[:1]

    ax.plot(
        angles,
        values,
        linewidth=2,
        label=f"Cluster {cluster}"
    )

    ax.fill(
        angles,
        values,
        alpha=0.08
    )


ax.set_xticks(
    angles[:-1]
)

ax.set_xticklabels(
    radar_features,
    fontsize=8
)

ax.set_title(
    "Customer Segment Radar Chart",
    pad=25
)

ax.legend(
    loc="upper right",
    bbox_to_anchor=(1.3, 1.1)
)

save_plot(
    "12_cluster_radar_chart.png"
)


# =====================================================================
# 45. BOX PLOTS ACROSS CLUSTERS
# =====================================================================

top_box_features = (
    feature_variability
    .head(8)
    .index
    .tolist()
)

fig, axes = plt.subplots(
    4,
    2,
    figsize=(15, 18)
)

axes = axes.flatten()

for i, feature in enumerate(
    top_box_features
):

    data_for_clusters = []

    for cluster in sorted(
        profile_data["Cluster"].unique()
    ):

        values = profile_data.loc[
            profile_data["Cluster"] == cluster,
            feature
        ].values

        data_for_clusters.append(
            values
        )

    axes[i].boxplot(
        data_for_clusters,
        labels=[
            f"C{x}"
            for x in sorted(
                profile_data[
                    "Cluster"
                ].unique()
            )
        ]
    )

    axes[i].set_title(
        feature
    )

    axes[i].set_xlabel(
        "Cluster"
    )

    axes[i].set_ylabel(
        "Value"
    )

plt.suptitle(
    "Feature Distributions Across SOM Clusters",
    fontsize=16,
    y=1.01
)

save_plot(
    "13_cluster_boxplots.png"
)


# =====================================================================
# 46. ANOVA STATISTICAL COMPARISON
# =====================================================================

anova_results = []

unique_clusters = sorted(
    profile_data["Cluster"].unique()
)

for feature in feature_columns:

    groups = []

    for cluster in unique_clusters:

        values = profile_data.loc[
            profile_data["Cluster"] == cluster,
            feature
        ].dropna().values

        groups.append(
            values
        )

    try:

        f_statistic, p_value = f_oneway(
            *groups
        )

    except Exception:

        f_statistic = np.nan
        p_value = np.nan

    anova_results.append({
        "Feature": feature,
        "F_Statistic": f_statistic,
        "P_Value": p_value,
        "Significant_At_0.05":
            bool(
                p_value < 0.05
            )
            if not np.isnan(p_value)
            else False
    })


anova_df = pd.DataFrame(
    anova_results
)

anova_df = anova_df.sort_values(
    "F_Statistic",
    ascending=False
)

anova_df.to_csv(
    DATA_DIR /
    "anova_cluster_comparison.csv",
    index=False
)

print("\nTop distinguishing features:")
print(
    anova_df.head(10)
)


# =====================================================================
# 47. ANOVA FEATURE IMPORTANCE GRAPH
# =====================================================================

top_anova = (
    anova_df
    .dropna()
    .head(10)
    .sort_values(
        "F_Statistic"
    )
)

plt.figure(
    figsize=(10, 7)
)

plt.barh(
    top_anova["Feature"],
    top_anova["F_Statistic"]
)

plt.xlabel(
    "ANOVA F-Statistic"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "Top Features Distinguishing Customer Segments"
)

save_plot(
    "14_anova_feature_importance.png"
)


# =====================================================================
# 48. DIRECT K-MEANS CUSTOMER CLUSTERING
# =====================================================================

direct_kmeans = KMeans(
    n_clusters=NUMBER_OF_CLUSTERS,
    random_state=RANDOM_STATE,
    n_init=20
)

kmeans_labels = direct_kmeans.fit_predict(
    X
)


# =====================================================================
# 49. SOM VS K-MEANS COMPARISON
# =====================================================================

som_silhouette = silhouette_score(
    X,
    customer_clusters
)

kmeans_silhouette = silhouette_score(
    X,
    kmeans_labels
)

som_db = davies_bouldin_score(
    X,
    customer_clusters
)

kmeans_db = davies_bouldin_score(
    X,
    kmeans_labels
)

ari = adjusted_rand_score(
    customer_clusters,
    kmeans_labels
)


comparison_df = pd.DataFrame({
    "Method": [
        "SOM + Neuron KMeans",
        "Direct KMeans"
    ],

    "Silhouette_Score": [
        som_silhouette,
        kmeans_silhouette
    ],

    "Davies_Bouldin_Index": [
        som_db,
        kmeans_db
    ]
})


comparison_df.to_csv(
    DATA_DIR /
    "som_vs_kmeans_comparison.csv",
    index=False
)


print("\nSOM vs K-Means:")
print(comparison_df)

print(
    "\nAdjusted Rand Index:",
    ari
)


# =====================================================================
# 50. SOM VS K-MEANS COMPARISON GRAPH
# =====================================================================

plt.figure(
    figsize=(8, 6)
)

plt.bar(
    comparison_df["Method"],
    comparison_df["Silhouette_Score"]
)

plt.ylabel(
    "Silhouette Score"
)

plt.title(
    "SOM-Based Clustering vs Direct K-Means"
)

plt.xticks(
    rotation=10
)

save_plot(
    "15_som_vs_kmeans_comparison.png"
)


# =====================================================================
# 51. SAVE K-MEANS ASSIGNMENTS
# =====================================================================

kmeans_results = pd.DataFrame({
    "CUST_ID": customer_ids.values,
    "KMEANS_CLUSTER":
        kmeans_labels + 1,
    "SOM_CLUSTER":
        customer_clusters + 1
})

kmeans_results.to_csv(
    DATA_DIR /
    "som_kmeans_customer_comparison.csv",
    index=False
)


# =====================================================================
# 52. IDENTIFY POTENTIAL AT-RISK CUSTOMERS
#
# Simple business-oriented rule:
# Lower purchases + lower payments + relatively high balance.
# =====================================================================

risk_df = results_df.copy()

risk_conditions = []

if "PURCHASES" in risk_df.columns:

    purchase_threshold = (
        risk_df["PURCHASES"]
        .quantile(0.25)
    )

    risk_conditions.append(
        risk_df["PURCHASES"]
        <= purchase_threshold
    )


if "PAYMENTS" in risk_df.columns:

    payment_threshold = (
        risk_df["PAYMENTS"]
        .quantile(0.25)
    )

    risk_conditions.append(
        risk_df["PAYMENTS"]
        <= payment_threshold
    )


if "BALANCE" in risk_df.columns:

    balance_threshold = (
        risk_df["BALANCE"]
        .quantile(0.75)
    )

    risk_conditions.append(
        risk_df["BALANCE"]
        >= balance_threshold
    )


if len(risk_conditions) > 0:

    risk_score = np.zeros(
        len(risk_df),
        dtype=int
    )

    for condition in risk_conditions:

        risk_score += condition.astype(int)

    risk_df[
        "RISK_SCORE"
    ] = risk_score

    risk_df[
        "POTENTIAL_AT_RISK"
    ] = (
        risk_score
        >= max(
            2,
            len(risk_conditions) - 1
        )
    )

else:

    risk_df[
        "RISK_SCORE"
    ] = 0

    risk_df[
        "POTENTIAL_AT_RISK"
    ] = False


risk_df.to_csv(
    DATA_DIR /
    "customer_risk_analysis.csv",
    index=False
)


# =====================================================================
# 53. AUTOMATIC CLUSTER CHARACTERISTICS
# =====================================================================

overall_means = capped_df.mean()

cluster_descriptions = {}

for cluster in cluster_means.index:

    cluster_values = (
        cluster_means.loc[cluster]
    )

    ratios = (
        cluster_values
        /
        (
            overall_means + epsilon
        )
    )

    high_features = (
        ratios
        .sort_values(
            ascending=False
        )
        .head(5)
        .index
        .tolist()
    )

    low_features = (
        ratios
        .sort_values()
        .head(5)
        .index
        .tolist()
    )

    cluster_descriptions[
        str(cluster)
    ] = {
        "high_features":
            high_features,

        "low_features":
            low_features
    }


# =====================================================================
# 54. BUSINESS RECOMMENDATION FUNCTION
# =====================================================================

def generate_recommendation(
    cluster_number
):

    means = cluster_means.loc[
        cluster_number
    ]

    recommendations = []

    if "PURCHASES" in means.index:

        if (
            means["PURCHASES"]
            >
            overall_means["PURCHASES"]
        ):

            recommendations.append(
                "Offer premium rewards, loyalty benefits "
                "and personalized high-value promotions."
            )

        else:

            recommendations.append(
                "Use purchase incentives, cashback campaigns "
                "and personalized spending offers."
            )


    if "CASH_ADVANCE" in means.index:

        if (
            means["CASH_ADVANCE"]
            >
            overall_means["CASH_ADVANCE"]
        ):

            recommendations.append(
                "Provide lower-cost repayment options and "
                "targeted balance-management products."
            )


    if (
        "CREDIT_LIMIT" in means.index
        and
        "PURCHASES" in means.index
    ):

        if (
            means["CREDIT_LIMIT"]
            >
            overall_means["CREDIT_LIMIT"]
            and
            means["PURCHASES"]
            >
            overall_means["PURCHASES"]
        ):

            recommendations.append(
                "Consider cross-selling premium cards, "
                "travel rewards and lifestyle products."
            )


    if "PAYMENTS" in means.index:

        if (
            means["PAYMENTS"]
            <
            overall_means["PAYMENTS"]
        ):

            recommendations.append(
                "Use payment reminders and retention "
                "campaigns to reduce potential disengagement."
            )


    if len(recommendations) == 0:

        recommendations.append(
            "Use personalized engagement campaigns based "
            "on the customer's dominant behavioral features."
        )

    return recommendations


# =====================================================================
# 55. CREATE CLUSTER PROFILE DOCUMENT
# =====================================================================

profile_file = (
    REPORT_DIR /
    "cluster_profiles.txt"
)

with open(
    profile_file,
    "w",
    encoding="utf-8"
) as file:

    file.write(
        "CUSTOMER SEGMENTATION USING "
        "SELF-ORGANIZING MAPS\n"
    )

    file.write(
        "=" * 70 + "\n\n"
    )

    file.write(
        f"Total customers: {len(df)}\n"
    )

    file.write(
        f"SOM map: {SOM_X} x {SOM_Y}\n"
    )

    file.write(
        f"Clusters: {NUMBER_OF_CLUSTERS}\n"
    )

    file.write(
        f"Training iterations: "
        f"{SOM_ITERATIONS}\n\n"
    )


    for cluster in cluster_means.index:

        count = int(
            (
                customer_clusters + 1
                == cluster
            ).sum()
        )

        percentage = (
            count /
            len(customer_clusters)
            * 100
        )

        file.write(
            f"\nCLUSTER {cluster}\n"
        )

        file.write(
            "-" * 40 + "\n"
        )

        file.write(
            f"Customers: {count}\n"
        )

        file.write(
            f"Percentage: "
            f"{percentage:.2f}%\n\n"
        )

        file.write(
            "Strongest characteristics:\n"
        )

        for feature in (
            cluster_descriptions[
                str(cluster)
            ]["high_features"]
        ):

            file.write(
                f"- High {feature}\n"
            )

        file.write(
            "\nRelatively low characteristics:\n"
        )

        for feature in (
            cluster_descriptions[
                str(cluster)
            ]["low_features"]
        ):

            file.write(
                f"- Low {feature}\n"
            )

        file.write(
            "\nBusiness recommendations:\n"
        )

        recommendations = (
            generate_recommendation(
                cluster
            )
        )

        for recommendation in recommendations:

            file.write(
                f"- {recommendation}\n"
            )

        file.write("\n")


print(
    "\nCluster profile document saved."
)


# =====================================================================
# 56. EXECUTIVE SUMMARY
# =====================================================================

largest_cluster = int(
    cluster_counts.idxmax()
)

largest_percentage = float(
    cluster_percentage.loc[
        largest_cluster
    ]
)

top_features = (
    anova_df
    .dropna()
    .head(5)["Feature"]
    .tolist()
)


executive_summary = f"""
CUSTOMER SEGMENTATION USING SELF-ORGANIZING MAPS
EXECUTIVE SUMMARY
============================================================

Dataset
-------
Total customers analysed: {len(df)}
Original attributes: {df.shape[1]}
Features used after engineering: {len(feature_columns)}

Methodology
-----------
The customer dataset was cleaned using median imputation for
missing numerical values. Extreme observations were identified
using the IQR method and capped rather than removed to retain
potentially important high-value customer behavior.

Additional behavioral ratio features were generated. Both
Min-Max normalization and z-score standardization were examined.

A {SOM_X} x {SOM_Y} Self-Organizing Map was trained using:

Learning rate: {SOM_LEARNING_RATE}
Initial neighborhood radius: {SOM_SIGMA}
Neighborhood function: Gaussian
Distance metric: Euclidean
Iterations: {SOM_ITERATIONS}
Initialization: PCA
Topology: Rectangular

Customer Segmentation
---------------------
The trained SOM neurons were grouped into
{NUMBER_OF_CLUSTERS} customer segments using K-Means.

Largest segment:
Cluster {largest_cluster}

Percentage of customers:
{largest_percentage:.2f}%

Important Segmentation Features
-------------------------------
{", ".join(top_features)}

Clustering Evaluation
---------------------
SOM Silhouette Score:
{som_silhouette:.4f}

Direct K-Means Silhouette Score:
{kmeans_silhouette:.4f}

SOM Davies-Bouldin Index:
{som_db:.4f}

Direct K-Means Davies-Bouldin Index:
{kmeans_db:.4f}

Adjusted Rand Index between SOM and K-Means:
{ari:.4f}

Business Interpretation
-----------------------
The SOM provides a two-dimensional representation of complex
customer behavior while preserving neighborhood relationships.
Customers with similar financial behavior tend to activate
nearby neurons.

The discovered customer groups can support:

1. Personalized marketing campaigns
2. Loyalty and reward programs
3. Customer retention strategies
4. Credit-product cross-selling
5. Identification of potentially inactive or risky customers
6. Premium-product targeting

At-Risk Customers
-----------------
Potentially at-risk customers were flagged using low purchase
activity, low payment activity and high balance behavior.
These customers can be targeted using retention campaigns,
payment reminders and personalized financial offers.

Conclusion
----------
Self-Organizing Maps provide both clustering and intuitive
visualization of high-dimensional customer behavior.

Unlike ordinary K-Means, the SOM maps customers onto an
organized two-dimensional neuron grid. This makes neighborhood
relationships, behavioral transitions, cluster boundaries and
outliers easier to visualize.
"""


with open(
    REPORT_DIR /
    "executive_summary.txt",
    "w",
    encoding="utf-8"
) as file:

    file.write(
        executive_summary
    )


print(
    "\nExecutive summary saved."
)


# =====================================================================
# 57. SAVE CONFIGURATION JSON
# =====================================================================

configuration = {

    "dataset":
        str(DATA_PATH),

    "som_grid": [
        SOM_X,
        SOM_Y
    ],

    "topology":
        "rectangular",

    "initial_learning_rate":
        SOM_LEARNING_RATE,

    "initial_sigma":
        SOM_SIGMA,

    "neighborhood_function":
        "gaussian",

    "distance_metric":
        "euclidean",

    "training_iterations":
        SOM_ITERATIONS,

    "initialization":
        "PCA",

    "number_of_clusters":
        NUMBER_OF_CLUSTERS,

    "features":
        feature_columns,

    "random_state":
        RANDOM_STATE,

    "final_quantization_error":
        float(
            error_history[-1]
        ),

    "som_silhouette_score":
        float(
            som_silhouette
        ),

    "kmeans_silhouette_score":
        float(
            kmeans_silhouette
        ),

    "adjusted_rand_index":
        float(
            ari
        )
}


with open(
    MODELS_DIR /
    "model_configuration.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        configuration,
        file,
        indent=4
    )


# =====================================================================
# 58. SAVE FEATURE NAMES
# =====================================================================

with open(
    MODELS_DIR /
    "feature_names.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        feature_columns,
        file,
        indent=4
    )


# =====================================================================
# 59. SAVE KMEANS MODEL
# =====================================================================

with open(
    MODELS_DIR /
    "som_neuron_kmeans.pkl",
    "wb"
) as file:

    pickle.dump(
        som_kmeans,
        file,
        protocol=pickle.HIGHEST_PROTOCOL
    )


# =====================================================================
# 60. CREATE README
# =====================================================================

readme = f"""
# Customer Segmentation Using Self-Organizing Maps

## Project Overview

This project performs customer segmentation using a
Self-Organizing Map (SOM), also known as a Kohonen Network.

The model maps high-dimensional credit-card customer behavior
onto a two-dimensional neuron grid while attempting to preserve
topological relationships.

## Dataset

Credit Card Customer Dataset

Customers analysed: {len(df)}

## Methodology

1. Data cleaning
2. Median missing-value imputation
3. IQR outlier detection
4. IQR-based outlier capping
5. Feature engineering
6. Min-Max normalization
7. Z-score standardization comparison
8. PCA dimensionality analysis
9. Self-Organizing Map training
10. SOM neuron clustering using K-Means
11. Customer mapping
12. Cluster profiling
13. ANOVA statistical analysis
14. Business interpretation
15. Comparison against direct K-Means

## SOM Configuration

Grid size: {SOM_X} x {SOM_Y}

Topology: Rectangular

Initialization: PCA-based

Learning rate: {SOM_LEARNING_RATE}

Neighborhood radius: {SOM_SIGMA}

Neighborhood function: Gaussian

Distance metric: Euclidean

Training iterations: {SOM_ITERATIONS}

## Number of Customer Segments

{NUMBER_OF_CLUSTERS}

## Important Outputs

### U-Matrix

`graphs/06_u_matrix.png`

The U-Matrix visualizes distances between neighboring SOM
neurons. Higher distances can indicate boundaries between
customer groups.

### Hit Map

`graphs/07_hit_map.png`

The hit map shows the number of customers mapped to each neuron.

### Cluster Map

`graphs/08_cluster_labeled_som_map.png`

Shows the customer segments identified by clustering the
trained SOM neurons.

### Radar Chart

`graphs/12_cluster_radar_chart.png`

Compares the behavioral profiles of the customer segments.

### ANOVA Analysis

`graphs/14_anova_feature_importance.png`

Shows features that differ strongly between customer segments.

### SOM vs K-Means

`graphs/15_som_vs_kmeans_comparison.png`

Compares SOM-based segmentation with conventional K-Means.

## Model Files

- `models/trained_som.pkl`
- `models/som_weights.npz`
- `models/minmax_scaler.pkl`
- `models/standard_scaler.pkl`
- `models/som_neuron_kmeans.pkl`
- `models/model_configuration.json`
- `models/feature_names.json`

## Data Outputs

- `customer_cluster_assignments.csv`
- `preprocessed_data_with_clusters.csv`
- `cluster_feature_means.csv`
- `cluster_feature_std.csv`
- `cluster_sizes.csv`
- `anova_cluster_comparison.csv`
- `customer_risk_analysis.csv`
- `som_vs_kmeans_comparison.csv`

## Business Applications

The customer segments can be used for:

- Targeted marketing
- Customer retention
- Premium product recommendations
- Loyalty campaigns
- Credit-product cross-selling
- Customer risk monitoring
- Personalized financial offers

## Conclusion

Self-Organizing Maps provide a useful approach for customer
segmentation because they combine unsupervised learning with
two-dimensional visualization.

The U-Matrix, component planes and hit maps provide additional
information about customer relationships that conventional
clustering algorithms such as K-Means do not directly provide.
"""


with open(
    RESULTS_DIR /
    "README.md",
    "w",
    encoding="utf-8"
) as file:

    file.write(
        readme.strip()
    )


# =====================================================================
# 61. CREATE RESULTS MANIFEST
# =====================================================================

manifest_rows = []

for file_path in RESULTS_DIR.rglob("*"):

    if file_path.is_file():

        size_mb = (
            file_path.stat().st_size
            /
            (1024 * 1024)
        )

        manifest_rows.append({
            "File":
                str(
                    file_path.relative_to(
                        RESULTS_DIR
                    )
                ),

            "Size_MB":
                round(
                    size_mb,
                    4
                )
        })


manifest_df = pd.DataFrame(
    manifest_rows
)

manifest_df.to_csv(
    RESULTS_DIR /
    "results_manifest.csv",
    index=False
)


# =====================================================================
# 62. CREATE FINAL ZIP
#
# ZIP contains results only, not original dataset.
# This keeps the submission small.
# =====================================================================

ZIP_PATH = (
    PROJECT_DIR /
    "SOM_Customer_Segmentation_Final.zip"
)

if ZIP_PATH.exists():
    ZIP_PATH.unlink()


MAX_SINGLE_FILE_MB = 20

with zipfile.ZipFile(
    ZIP_PATH,
    "w",
    compression=zipfile.ZIP_DEFLATED,
    compresslevel=9
) as zip_file:

    for file_path in RESULTS_DIR.rglob("*"):

        if not file_path.is_file():
            continue

        file_size_mb = (
            file_path.stat().st_size
            /
            (1024 * 1024)
        )

        if file_size_mb > MAX_SINGLE_FILE_MB:

            print(
                "Skipping large file:",
                file_path.name,
                f"({file_size_mb:.2f} MB)"
            )

            continue

        archive_name = (
            file_path.relative_to(
                PROJECT_DIR
            )
        )

        zip_file.write(
            file_path,
            arcname=archive_name
        )


# =====================================================================
# 63. CHECK FINAL ZIP SIZE
# =====================================================================

zip_size_mb = (
    ZIP_PATH.stat().st_size
    /
    (1024 * 1024)
)

print("\n" + "=" * 70)

print(
    "FINAL ZIP CREATED"
)

print("=" * 70)

print(
    "\nZIP:",
    ZIP_PATH
)

print(
    f"\nZIP size: "
    f"{zip_size_mb:.2f} MB"
)


if zip_size_mb < 50:

    print(
        "\nSUCCESS: Final ZIP is under 50 MB."
    )

else:

    print(
        "\nWARNING: ZIP exceeds 50 MB."
    )

    print(
        "Consider reducing graph DPI or "
        "excluding trained_som.pkl."
    )


# =====================================================================
# 64. DISPLAY FINAL SUMMARY
# =====================================================================

print("\n" + "=" * 70)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 70)

print(
    f"""
Customers analysed:
{len(df)}

Features used:
{len(feature_columns)}

SOM:
{SOM_X} x {SOM_Y}

Training iterations:
{SOM_ITERATIONS}

Clusters:
{NUMBER_OF_CLUSTERS}

Final Quantization Error:
{error_history[-1]:.6f}

SOM Silhouette Score:
{som_silhouette:.4f}

K-Means Silhouette Score:
{kmeans_silhouette:.4f}

Adjusted Rand Index:
{ari:.4f}

Results directory:
{RESULTS_DIR}

Final ZIP:
{ZIP_PATH}

ZIP Size:
{zip_size_mb:.2f} MB
"""
)


# =====================================================================
# 65. LIST GENERATED FILES
# =====================================================================

print("\nGENERATED FILES")
print("-" * 70)

for file_path in sorted(
    RESULTS_DIR.rglob("*")
):

    if file_path.is_file():

        size_kb = (
            file_path.stat().st_size
            /
            1024
        )

        print(
            f"{file_path.relative_to(RESULTS_DIR)} "
            f"- {size_kb:.2f} KB"
        )


print("\nDone.")
```
